# Logistic Regression - Theoretical Questions and Answers

### 1. What is Logistic Regression, and how does it differ from Linear Regression

**Answer:** Logistic Regression is used for classification problems where the output is categorical. Unlike Linear Regression which outputs continuous values, Logistic Regression predicts the probability that a given input point belongs to a certain class using the sigmoid function.

### 2. What is the mathematical equation of Logistic Regression

**Answer:** The mathematical form is: P(Y=1|X) = 1 / (1 + exp(-(β₀ + β₁X₁ + ... + βₙXₙ)))

### 3. Why do we use the Sigmoid function in Logistic Regression

**Answer:** The sigmoid function squashes input values into a range between 0 and 1, which makes it suitable for modeling probabilities.

### 4. What is the cost function of Logistic Regression

**Answer:** Logistic Regression uses the binary cross-entropy (log loss) cost function: -[y*log(p) + (1−y)*log(1−p)]

### 5. What is Regularization in Logistic Regression? Why is it needed

**Answer:** Regularization adds a penalty to the loss function to prevent overfitting. It keeps the model weights small and generalizes better to unseen data.

### 6. Explain the difference between Lasso, Ridge, and Elastic Net regression

**Answer:** Lasso uses L1 penalty and can shrink coefficients to zero (feature selection), Ridge uses L2 penalty which shrinks coefficients but never to zero. Elastic Net combines both L1 and L2 penalties.

### 7. When should we use Elastic Net instead of Lasso or Ridge

**Answer:** Elastic Net is preferred when there are multiple features correlated with each other, and we want the benefits of both L1 and L2 regularization.

### 8. What is the impact of the regularization parameter (λ) in Logistic Regression

**Answer:** A larger λ increases the penalty on large coefficients, making the model simpler. A smaller λ allows more flexibility but may lead to overfitting.

### 9. What are the key assumptions of Logistic Regression

**Answer:** Assumptions include linear relationship between logit and predictors, no multicollinearity, large sample size, and independence of errors.

### 10. What are some alternatives to Logistic Regression for classification tasks

**Answer:** Alternatives include Decision Trees, Random Forests, SVMs, KNN, Naive Bayes, and Neural Networks.

### 11. What are Classification Evaluation Metrics

**Answer:** Accuracy, Precision, Recall, F1-score, ROC-AUC, Log Loss, and Confusion Matrix are common metrics.

### 12. How does class imbalance affect Logistic Regression

**Answer:** Class imbalance can cause the model to bias predictions toward the majority class. It reduces effectiveness of metrics like accuracy.

### 13. What is Hyperparameter Tuning in Logistic Regression

**Answer:** Hyperparameter tuning involves finding the best settings like regularization strength (C), penalty type, or solver using methods like GridSearchCV.

### 14. What are different solvers in Logistic Regression? Which one should be used

**Answer:** Common solvers are 'liblinear', 'saga', 'lbfgs', and 'newton-cg'. 'liblinear' is good for small datasets and L1, 'saga' supports all penalties and large datasets.

### 15. How is Logistic Regression extended for multiclass classification

**Answer:** It is extended using One-vs-Rest (OvR) or Softmax (multinomial) strategies.

### 16. What are the advantages and disadvantages of Logistic Regression

**Answer:** Advantages: simple, interpretable, fast. Disadvantages: assumes linear decision boundary, poor performance with highly non-linear relationships.

### 17. What are some use cases of Logistic Regression

**Answer:** Use cases include spam detection, credit scoring, medical diagnosis, and churn prediction.

### 18. What is the difference between Softmax Regression and Logistic Regression

**Answer:** Logistic Regression is for binary classification, Softmax Regression is its generalization for multiclass classification.

### 19. How do we choose between One-vs-Rest (OvR) and Softmax for multiclass classification

**Answer:** OvR trains one classifier per class, Softmax trains a single model for all classes. OvR is preferred for sparse classes, Softmax for mutually exclusive classes.

### 20. How do we interpret coefficients in Logistic Regression?

**Answer:** Each coefficient represents the log odds change of the target variable for a one-unit increase in the predictor, holding other variables constant.
